Problem 1:
A fair coin is tossed 100 times. Calculate:
*   Number of heads
*   Number of tails
*   Experimental probability of head
*   Experimental Probability of tail

```python
import random   # module used to generate random values

# initialize counters for head and tail to zero
heads = 0
tails = 0

# for 100 iterations.
for i in range(100):
  toss = random.choice(["H", "T"])  # randomly selects one item

# check whether the toss in head
  if toss == "H":
     heads += 1    # increase the head count by 1

# if toss is tail
  else:
    tails += 1     # increase the tail count by 1

# display the output
print("Number of Heads:" , heads)
print("Number of Tails:" , tails)

print("Experimental probability of head:" , heads/100)
print("Experimental probability of tails:" , tails/100)
```

Problem 2:

Check effect of the the number of trails

Repeat the coin-toss experiment for n=10, 100, 1000, 10000

```python
import random  # module used to generate random values

def coin_toss(n):

# initialize the count of heads and tails to zero
  heads = 0
  tails = 0

# repeat the coin toss n times
  for i in range(n):
    toss = random.choice(["H", "T"])  # randomly selects one item

# check whether the toss in head
    if toss == "H":
      heads += 1     # increase the head count by 1

# if toss is tail
    else:
      tails +=1      # increase the tail count by 1

# display the output
  print("Number of trials:",n)
  print("Heads:", heads)
  print("Tails:", tails)
  print("Probability of Head:", heads / n)
  print("probability of Tail:", tails / n)
  print()

coin_toss(10)
coin_toss(100)
coin_toss(1000)
coin_toss(10000)
```

Problem 3:

Rolling a dice: A fair dice is rolled 6000 times.

Calculate the experimental probability of each outcomes: S = {1,2,3,4,5,6}

```python
import random  # module used to generate random values

# initialize count for each dice outcome
count = [0,0,0,0,0,0]

# roll the dice for 6000 times
for i in range(6000):
  roll = random.randint(1,6)   # generates random number between 1 to 6
  count[roll-1] += 1           # increases the count of rolled outcome

# display the results for each outcome
for i in range(6):
  print("Outcome", i+1)
  print("Count:", count[i])
  print("Experimental Probability:", count[i] / 6000)
  print()
```

Problem 4:

Addition rule for disjoint events

A dice is rolled 10000 times.
Define: A = getting 1, B = getting 6, check the events are joint or disjoint.

Verify: P(A∪B)=P(A)+P(B)

```python
import random   # module used to generate random values

# initialize counters for events A, B, and A_union_B
A = 0
B = 0
A_union_B = 0

# roll the dice for 10000 times
for i in range(10000):
  roll = random.randint(1, 6)  # generates a random number between 1 and 6

# checks if the outcome is 1 for event A
  if roll == 1:
    A += 1       # increase the count of event A by 1

# checks if the outcome is 6 for event B
  if roll == 6:
    B += 1       # increase the count of event B by 1

# checks if the outcome is either 1 or 6
  if roll == 1 or roll == 6:
    A_union_B += 1   # increase the count of event A_union_B by 1

# calculate experimental probability
P_A = A / 10000
P_B = B / 10000
P_A_union_B = A_union_B / 10000

# display the calculated probabilities
print("P(A):", P_A)
print("P(B):", P_B)
print("P(A∪B):", P_A_union_B)
print()

# display the number of times each event occurred
print("Number of times A occurred:", A)
print("Number of times B occurred:", B)
print("Number of times A union B occurred:", A_union_B)
print()

# calculate and display results
print("P(A) + P(B):", P_A + P_B)
print()

# verification
if abs(P_A_union_B - (P_A +P_B)) < 0.01:
  print("Verified: P(A union B) = P(A) + P(B)")
else:
  print("Verification failed")
```

Problem 5:

Addition rule for overlapping events

For one dice roll, define: A = getting an even number A={2,4,6} and B=getting a number greater than 3 B={4,5,6}
Verify: P(A∪B)=P(A)+P(B)−P(A∩B)

```python
import random   # module used to generate random values

trials = 10000  # sets the total number of trials to 10000

# initialize counters for events A, B, A_union_B, and A_intersection_B
A = 0
B = 0
A_union_B = 0
A_intersection_B = 0

# roll the dice for 10000 times
for i in range(trials):
  roll = random.randint(1, 6)  # generates a random number between 1 and 6

# checks if outcome belongs to event A
  if roll in [2, 4, 6]:
    A += 1   # increase the count of event A by 1

# checks if outcome belongs to event B
  if roll in [4, 5, 6]:
    B += 1   # increase the count of event B by 1

# checks if outcome belongs to both A and B
  if roll in [4,6]:
    A_intersection_B += 1  # increase the count of intersection by 1

# checks if outcome belongs to A or B
  if roll in [2, 4, 5, 6]:
    A_union_B += 1   # increase the count of union by 1

# calculates experimental probabilities
P_A = A / trials
P_B = B / trials
P_union = A_union_B / trials
P_intersection = A_intersection_B / trials

# display the calculated probabilities
print("P(A):", P_A)
print("P(B):", P_B)
print()
print("P(A union B):", P_union)
print("P(A intersection B):", P_intersection)
print()

# verify the addition rule of probability
print("P(A) + P(B) - P(A intersection B):", P_A + P_B - P_intersection)
print()

if abs(P_union - (P_A + P_B - P_intersection)) < 0.01:
  print("Verified")
else:
  print("Verification failed")
```

Problem 6:

Multiplication rule for independent events.

A coin is tossed and a dice is rolled 10,000 times.
Define :

*   Obtaining Head
*   Obtaining an even number

Verify : P(A∩B)=P(A)P(B)



```python
import random  # module used to generate random values

trials = 10000  # sets the total number of trials to 10000

# initialize counters for events A, B, and A_intersection_B
A = 0
B = 0
A_intersection_B = 0

# repeat the experiment for 10000 times
for i in range(trials):

  coin = random.choice(["H", "T"])  # randomly selects one item
  dice = random.randint(1, 6)  # generates a random number between 1 and 6

  # checks if the coin result is Head
  if coin == "H":
    A += 1   # increase the count of event A by 1

  # checks if the dice result is even number
  if dice % 2 == 0:
    B += 1  # increase the count of event B by 1

  # checks if both head and even number occurred
  if coin == "H" and dice % 2 == 0:
    A_intersection_B += 1  # increase the count of A_intersection_B by 1

# calculate experimental probabilities
P_A = A / trials
P_B = B / trials
P_intersection = A_intersection_B / trials

# display the number of trials and event counts
print("Number of trials:", trials)
print("A(Head):", A)
print("B(Even number):", B)
print("A intersection B:", A_intersection_B)
print()

# display the experimental probabilities
print("P(A):", P_A)
print("P(B):", P_B)
print("P(A intersection B):", P_intersection)
print()

# verify the multiplication rule for independent events
print("P(A) * P(B):", P_A * P_B)

if abs(P_intersection - (P_A * P_B)) < 0.01:
  print("Verified")
else:
  print("Verification failed")
```
