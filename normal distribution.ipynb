```python
import numpy as np
```

```python
# Step 1: Generate 10,000 random values
np.random.seed(42)

population = np.random.normal(
    loc=50,
    scale=10,
    size=10000
)
```

```python
# Step 2: Calculate population statistics
population_mean = np.mean(population)
population_std = np.std(population)

print("Population Mean:", population_mean)
print("Population Standard Deviation:", population_std)
```

```python
# Step 3: Randomly select 30 samples
sample = np.random.choice(population, size=30, replace=False)
```

```python
# Step 4: Calculate sample statistics
sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)
```

```python
# Step 5: Calculate Standard Error
standard_error = sample_std / np.sqrt(30)
```

```python
print("\nSample of 30 values:")
print(sample)
```

```python
print("\nSample Mean:", sample_mean)
print("Sample Standard Deviation:", sample_std)
print("Standard Error:", standard_error)
```
