```python
import matplotlib.pyplot as plt

# number of people entered by the user
n = int(input("Enter the number of people:"))

days = 365

# probability that no two people share a birthday
p_no_shared = 1.0

for person in range(n):
  p_no_shared = p_no_shared * (days - person) / days

# probability that at least two people share a birthday
p_shared = 1 - p_no_shared

print("probability of at least two people sharing a birthday:", round(p_shared, 4))

# calculate probabilities for 1 to 30 people
people = range(1, 31)

probabilities = []

for n in people:
  p_no_shared = 1.0


  for person in range(n):
    p_no_shared = p_no_shared * (days - person) / days

  p_shared = 1 - p_no_shared

  probabilities.append(p_shared)

# plot the graph
plt.plot(people, probabilities, marker='o')

plt.xlabel("Number of people")
plt.ylabel("Probability of shared birthday")
plt.title("Birthday Problem")

plt.grid(True)
plt.show()
```
